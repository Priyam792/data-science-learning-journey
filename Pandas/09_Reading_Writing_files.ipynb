## CSV Files

```python
import pandas as pd
```

```python
df = pd.read_csv("data.csv") # Read CSV
```

```python
df
```

```python
df = df[df['IMDb']>7.0]
df
```

```python
df.to_csv("data_updated.csv", index=False) #index= False means to remove the index (0,2,5,6,11) while writing
```

```python
df1 = pd.read_excel("students_data.xlsx")
```

```python
df1
```

```python
df1 = df1[df1['Marks']>80]
df1
```

### saving process is same as csv ***.to_excel***

## Multiple sheets ..

```python
with pd.ExcelWriter("actor_students_data_merged.xlsx") as writer:
    df.to_excel(writer, sheet_name="Actors", index=False)
    df1.to_excel(writer, sheet_name="Students", index=False)
```
