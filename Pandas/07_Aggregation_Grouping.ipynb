# ***Aggregation & Grouping***
Grouping and aggregating helps you summarize your data — like answering:
```
> “What’s the average salary per department?”
>“How many users joined the Gym per month?”
```

````python
import pandas as pd
````

````python
df = pd.DataFrame({
"Department": ["HR", "HR", "IT", "IT", "Marketing", "Marketing", "Sales", "Sales"],
"Team": ["A", "A", "B", "B", "C", "C", "D", "D"],
"Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
"Salary": [85, 90, 78, 85, 92, 88, 75, 80],
"Age": [23, 25, 30, 22, 28, 26, 21, 27],
"JoinDate": pd.to_datetime([
"2020-01-10", "2020-02-15", "2021-03-20", "2021-04-10",
"2020-05-30", "2020-06-25", "2021-07-15", "2021-08-01"
])
})
````

````python
df
````

````python
df.groupby("Department")["Salary"].mean() #it gives the mean of salary of all the department
````

## Common Aggregation Functions

````python
df.groupby("Department")["Salary"].sum() # gives the sum of the salary of all department
````

````python
df.groupby("Team")["Salary"].count() # give the no. of entiries for every team
````

````python
df.groupby("Department")["Salary"].min() # minimum salary of every department
````

````python
df.groupby("Team")["Salary"].max() # maximum salary of every team
````

````python
df.groupby(["Team", "Gender"])["Salary"].mean()
````

## Custom Aggregations with .agg() / .aggregate() both are exactly same -- they are alias for the same method . 

````python
df.groupby("Team")["Salary"].agg(["mean", "max", "min"])
````

````python
# Custom names 
df.groupby("Team")["Salary"].agg(
    Avg_salary = "mean",
    max_salary = "max",
    min_salary = "min")
````

````python
# Apply different functions at different columns
df.groupby("Team").agg({
"Salary": "mean",
"Age": "max"
})
````

## Transform vs Aggregate vs Filter
---


````python
df["Team Avg"] = df.groupby("Team")["Salary"].transform("mean") # makes a new row .
df
````

````python
df.groupby("Team").filter(lambda x: x["Salary"].mean() > 85) # it filters the transformed row
````

# Summary
---
```
• .groupby() helps you summarize large datasets by category
• Use mean() , sum() , count() , .agg() for custom metrics
• .transform() adds values back to original rows
• .filter() keeps only groups that meet conditions
```
