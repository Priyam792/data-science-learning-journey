# Merging & Joining Data
---
Often, data is split across multiple tables or files. Pandas lets you combine them just like SQL — or even more flexibly!

```python
import pandas as pd
```

```python
employees = pd.DataFrame({
"EmpID": [1, 2, 3],
"Name": ["Alice", "Bob", "Charlie"],
"DeptID": [10, 20, 30]
})
departments = pd.DataFrame({
"DeptID": [10, 20, 40],
"DeptName": ["HR", "Engineering", "Marketing"]
})
```

```python
employees
```

```python
departments
```

## Merge Like SQL: pd.merge()

```python
pd.merge(employees, departments , on= "DeptID") # Inner Joint (By default)
# on means common column (IF there is only one common column then pandas automantically takes this )
```

```python
pd.merge(employees, departments , on= "DeptID" , how= "right") # Right Joint using how=
# Takes all the column of right and common from left .. here right is departments cause it written in left
```

```python
pd.merge(employees, departments , on= "DeptID" , how= "left") # Left Joint using how=
```

```python
pd.merge(employees, departments , on= "DeptID" , how= "outer") # Outer Joint using how=
```

## Concatenating DataFrames
---
Use pd.concat() to stack datasets either vertically or horizontally.

```python
df1 = pd.DataFrame({"Name": ["Alice", "Bob"]})
df2 = pd.DataFrame({"Name": ["Charlie", "David"]})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2]) # Vertical by default
```

```python
pd.concat([df1,df2],axis=1) # Horizontal by using axis= 1
```
