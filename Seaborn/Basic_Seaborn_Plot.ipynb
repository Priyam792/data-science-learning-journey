# Why Seaborn?
- Matplotlib is powerful but very low-level.
- Seaborn adds high-level features like automatic styling, themes, color palettes, and dataframe integration.
- Seaborn comes with built in Datasets
- Makes complex plots (like boxplots, violin plots, heatmaps, pairplots) very easy.
# In short:
- Less code
- Better-looking graphs
- Easy handling of DataFrames (like from pandas)

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import numpy as np
x = np.array([0, 2, 3, 4, 5, 6, 7, 8, 9 ,10, 60])
y = np.sin(x)
```

```python
sns.lineplot(x=x,y=y)
plt.title("Line Plot")
plt.xlabel('x')
plt.ylabel('y')
```

- Seaborn comes with built-in example datasets.
- These are small real-world datasets like restaurant tips, flight passenger counts, iris flower measurements, etc.
- They are mainly used for practice, examples, and learning plotting techniques without needing to manually download any data.

```python
# how to see all the availabel datasets
print(sns.get_dataset_names())
```

```python
penguins = sns.load_dataset('penguins')
penguins
```

```python
penguins['island'].unique()
```

# Line plot

```python
sns.lineplot(x='flipper_length_mm' , y= 'body_mass_g' , data = penguins , color = 'red')
plt.title("PENGUIN")
```

# Scatter Plot

```python
sns.scatterplot(x='flipper_length_mm' , y= 'body_mass_g' , data = penguins , color = 'skyblue')
plt.title("PENGUIN")
```

# Bar Plot

```python
sns.barplot(x='species' , y= 'body_mass_g' , data = penguins , color = 'orange' , errorbar= None) #errorbar=False is to remove the vertial line on every bar of the graph
plt.title("PENGUIN")
plt.tight_layout()
```

# Box Plot
- Boxplots show distributions, medians, and outliers in one simple plot.

```python
sns.boxplot(x='species' , y= 'body_mass_g' , data = penguins , color = 'green')
plt.title("PENGUIN")
```

```python
flights = sns.load_dataset('flights')
pivot_table = flights.pivot(index = "month",columns = "year",values = "passengers")
sns.heatmap(pivot_table, annot=True, fmt="d", cmap="YlGnBu")
plt.title('Heatmap of Passengers')
```

```python
sns.scatterplot(
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species",
    data=penguins
) #hue seprates the different value by different colors here it diffrentiate on species 
```
