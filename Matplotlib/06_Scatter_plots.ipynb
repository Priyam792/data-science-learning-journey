# Scatter Plot in Matplotlib
## Scatter plots are used to show the relationship between two variables. Let's assume we are plotting the study hours vs. exam scores for a group of students.

```python
import matplotlib.pyplot as plt
```

```python
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [80, 41, 90, 15, 61, 22, 7, 75, 34]
```

```python
plt.scatter(study_hours, exam_scores)
plt.title('Study Hours vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
```

## Adding Color and Size

```python
sizes = [score*2 for score in exam_scores]
colors = ['green' if score > 70 else 'yellow' if score > 50 else 'red' for score in exam_scores ]
plt.scatter(study_hours, exam_scores ,edgecolors='black', s = sizes , c = colors , alpha = 0.6 )
plt.colorbar(label= exam_scores)
plt.title('Study Hours')
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True)
```

## Using a Heatmap

```python
plt.figure(figsize=(8, 6))
plt.scatter(study_hours, exam_scores, c= exam_scores ,s= sizes, cmap = 'magma')
plt.colorbar(label= 'Exam Scores')
plt.title('Study Hours vs Exam Scores')
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True)
```

## Adding Annotations

```python
plt.figure(figsize=(8, 6))
plt.scatter(study_hours, exam_scores, c= exam_scores ,s= sizes, cmap = 'magma')
plt.colorbar(label= 'Exam Scores')
for i in range(len(study_hours)):
    plt.annotate(f's{i+1}',(study_hours[i] + 0.2, exam_scores[i] + 0.1))
plt.title('Study Hours vs Exam Scores')
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.margins(x=0.1,y=0.1)
plt.grid(True)
```

## Multiple Groups in One Plot

```python
class_a_hours = [2, 3, 4, 5]
class_a_scores = [45, 55, 65, 85]
class_b_hours = [1, 4, 5, 2, 3]
class_b_scores = [40, 50, 60, 70, 90]
plt.scatter(class_a_hours, class_a_scores, label='Class A', color='blue')
plt.scatter(class_b_hours, class_b_scores, label='Class B', color='orange')
plt.title('Scatter Plot: Class A vs Class B')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.legend()
plt.grid(True)
```
